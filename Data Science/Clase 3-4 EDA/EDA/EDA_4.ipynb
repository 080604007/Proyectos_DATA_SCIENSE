```python
#Crear un nuevo libro con los datos de los libros:
#2.Cargar los datos a un data frame:
#3. head(), info(), describe().
#4.Sacar graficas por raiting.
#5. Sacar graficas por precio_gbp.

#1.Cargar Librerias

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

##Cargar el dataset
df = pd.read_csv("libros_toscrape_pag1.csv")
print("Datos cargados con éxito")
```

```python
df.head() #muestra las primeras 5
df.info() #informacion del dataset
df.describe() #estadisticas descriptivas del dataset
```

```python
#Configuracion estetica de las graficas
import seaborn as sns
sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"]=(10,6)

#Variable objetivo
sns.countplot(x="rating", data=df)
plt.title("Distribucion de ratings")
plt.show()
```

```python
#Variable precio_gbp.
sns.countplot(x="precio_gbp", data=df)
plt.title("Distribucion de precios en GBP")
plt.show()
```
